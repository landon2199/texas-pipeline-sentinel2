```python
import arcpy

# -----------------------------
# USER SETTINGS
# -----------------------------
input_fc = r"C:\391 zoom lesson testing\Extracted 2 data\391 compiled october 31\ArcProject (Download Here)\391 project compiled\ArcProject (Download Here)\391 Research\391 Research Phase 1\391 Research Phase 1.gdb\hydrocarbonsliquid_DiameterClass100mBuffer_EcoIntersect_SimplifiedPolygon"
csv_path = r"C:\391 zoom lesson testing\Extracted 2 data\391 compiled october 31\Hydrocarbon_Diameter_eco_100m\combined_all_regions_underscored_masterset.csv"
output_gdb = r"C:\391 zoom lesson testing\Extracted 2 data\391 compiled october 31\ArcProject (Download Here)\391 project compiled\ArcProject (Download Here)\391 Research\391 Research Phase 1\391 Research Phase 1.gdb"
gdb_table_name = "CSV_JoinTablehydrocarbon_100m"
join_field_fc = "PipelineEcoID"       # Key in feature class
join_field_csv = "PipelineEcoID"      # Key in CSV
base_name = "ExportChunk"
merged_fc = "MergedLayer"
chunk_size = 200000  # Adjust if needed

# -----------------------------
# ENVIRONMENT SETTINGS
# -----------------------------
arcpy.env.overwriteOutput = True
arcpy.env.maintainSpatialIndex = False  # Disable spatial index during export for speed

# -----------------------------
# STEP 1: Convert CSV to GDB Table
# -----------------------------
print("Converting CSV to GDB table...")
arcpy.TableToTable_conversion(csv_path, output_gdb, gdb_table_name)
print(f"✅ CSV converted to GDB table: {gdb_table_name}")

# -----------------------------
# STEP 2: Permanently Join Fields
# -----------------------------
print("Joining fields permanently...")
arcpy.management.JoinField(input_fc, join_field_fc, f"{output_gdb}\\{gdb_table_name}", join_field_csv)
print("✅ Join completed. NDVI and NDWI fields are now part of the feature class.")

# -----------------------------
# STEP 3: Detect ObjectID Field
# -----------------------------
fields = [f.name for f in arcpy.ListFields(input_fc)]
oid_field = next((f for f in fields if f.upper().endswith("OBJECTID")), None)

if not oid_field:
    raise ValueError("Could not find ObjectID field. Check your layer fields.")

print(f"Using ObjectID field: {oid_field}")

# -----------------------------
# STEP 4: Get ObjectID Range
# -----------------------------
oid_values = [row[0] for row in arcpy.da.SearchCursor(input_fc, oid_field)]
min_oid, max_oid = min(oid_values), max(oid_values)

print(f"ObjectID range: {min_oid} to {max_oid}")

# -----------------------------
# STEP 5: Export in Chunks
# -----------------------------
for start in range(min_oid, max_oid + 1, chunk_size):
    end = start + chunk_size - 1
    where_clause = f'"{oid_field}" >= {start} AND "{oid_field}" <= {end}'  # FIXED: Removed HTML entities
    out_name = f"{base_name}_{start}_{end}"
    print(f"Exporting {out_name}...")
    arcpy.FeatureClassToFeatureClass_conversion(input_fc, output_gdb, out_name, where_clause)

print("✅ All chunks exported successfully!")

# -----------------------------
# STEP 6: Merge Chunks
# -----------------------------
arcpy.env.workspace = output_gdb
chunks = arcpy.ListFeatureClasses(f"{base_name}_*")
print(f"Merging {len(chunks)} chunks into {merged_fc}...")
arcpy.Merge_management(chunks, merged_fc)

print("✅ Merge completed successfully!")

# -----------------------------
# STEP 7: Add Spatial Index
# -----------------------------
print("Adding spatial index for performance...")
arcpy.AddSpatialIndex_management(f"{output_gdb}\\{merged_fc}")

print("✅ Spatial index added. Final feature class ready for symbology!")
```
